```python
documentary=input("Enter your documentary name: ")
```

```python
!pip install wikipedia
import wikipedia

page = wikipedia.page(documentary)

text = page.content

print(text)
```

```python
chunk_size = 500

chunks = [
    text[i:i + chunk_size]
    for i in range(0, len(text), chunk_size)
]

print("Total chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i+1} | Size: {len(chunk)} ---")
    print(chunk)
```

```python
!pip install sentence-transformers
from sentence_transformers import SentenceTransformer
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
doc=chunks
```

```python
embeddings = embedding_model.encode(doc).tolist()
```

```python
!pip install chromadb

import chromadb
chroma_client = chromadb.Client()

collection = chroma_client.get_or_create_collection(name="Documentary_details")
```

```python
collection.add(
    documents=doc,
    embeddings=embeddings,
    ids=[f"id{i}" for i in range(len(doc))]
)
```

```python
def retrieve(query, top_k=2):
    query_embedding = embedding_model.encode([query]).tolist()

    results = collection.query(
        query_embeddings=query_embedding,
        n_results=top_k
    )

    return results["documents"][0]
```

```python
import gradio as gr
import wikipedia
import chromadb
from sentence_transformers import SentenceTransformer
from groq import Groq

# Embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# Chroma
chroma_client = chromadb.Client()

# Groq
client = Groq(api_key='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')


def answer_question(documentary, question):

    try:

        # -------------------------
        # 1. Get Wikipedia page
        # -------------------------

        page = wikipedia.page(
            documentary,
            auto_suggest=True
        )

        text = page.content


        # -------------------------
        # 2. Create chunks
        # -------------------------

        chunks = []

        for i in range(0, len(text), 500):

            chunks.append(
                text[i:i + 500]
            )


        # -------------------------
        # 3. Delete old collection
        # -------------------------

        try:

            chroma_client.delete_collection(
                "documentary_details"
            )

        except:

            pass


        # -------------------------
        # 4. Create collection
        # -------------------------

        collection = chroma_client.create_collection(
            name="documentary_details"
        )


        # -------------------------
        # 5. Create embeddings
        # -------------------------

        embeddings = embedding_model.encode(
            chunks
        ).tolist()


        # -------------------------
        # 6. Store chunks
        # -------------------------

        collection.add(

            documents=chunks,

            embeddings=embeddings,

            ids=[
                "doc" + str(i)
                for i in range(len(chunks))
            ]
        )


        # -------------------------
        # 7. Question embedding
        # -------------------------

        question_embedding = embedding_model.encode(
            [question]
        ).tolist()


        # -------------------------
        # 8. Search
        # -------------------------

        results = collection.query(

            query_embeddings=question_embedding,

            n_results=3
        )


        context = "\n\n".join(
            results["documents"][0]
        )


        # -------------------------
        # 9. Ask Groq
        # -------------------------

        prompt = f"""
You are a documentary question answering assistant.

Use ONLY the information given in the context.

Context:
{context}

Question:
{question}

Answer the question clearly and simply.

If the answer is not available in the context, say:

"This documentary does not contain the information about your question."
"""


        response = client.chat.completions.create(

            model="openai/gpt-oss-120b",

            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ],

            temperature=0.2
        )


        # -------------------------
        # 10. Return answer
        # -------------------------

        answer = response.choices[0].message.content

        return answer


    except Exception as e:

        return "Error: " + str(e)


# ==================================
# GRADIO
# ==================================

demo = gr.Interface(

    fn=answer_question,

    inputs=[

        gr.Textbox(
            label="Documentary Name",
            placeholder="Example: The Rescue"
        ),

        gr.Textbox(
            label="Question",
            placeholder="Example: What happened during the rescue?"
        )

    ],

    outputs=gr.Textbox(
        label="Answer"
    ),

    title="🎬 Documentary Q&A",

    description="Enter a documentary name and ask a question."
)


demo.launch(
    share=True
)
```
