```python
SUB=input('ENTER A SUBJECT: ')
LEV=input('ENTER A CURRENT LEVEL: ')
TIME=input('ENTER AVAILABLE TIME: ')
DUR=input('ENTER DURATION: ')
GOAL=input('ENTER GOAL: ')
```

```python
from langchain_core.prompts import PromptTemplate

prompt1 = PromptTemplate(
    input_variables=["SUB",'LEV','TIME','DUR','GOAL'],
    
    template="""
You are an expert learning strategist and curriculum designer.

Create a **detailed, practical, and personalized learning roadmap** based on the following information:

**Subject:** {SUB}
**Current Level:** {LEV}
**Available Study Time:** {TIME}
**Total Duration:** {DUR}
**Goal:** {GOAL}

Design the roadmap specifically for this learner. Do not give a generic curriculum.

Include the following sections:

### 1. Learning Strategy

aseBriefly explain how the learner should approach the subject bd on their current level, available time, duration, and goal.

### 2. Roadmap Overview

Divide the entire duration into logical phases or weeks.

For each phase, provide:

* Topics to learn
* Concepts to understand
* Practical skills to develop
* Expected outcome

### 3. Detailed Weekly Plan

For every week, provide:

* What to learn
* What to practice
* Recommended daily time allocation
* Exercises or activities
* Mini-projects/tasks
* Expected learning outcome

Make sure the workload fits within the available study time.

### 4. Practical Projects

Suggest progressively difficult projects that help the learner apply what they learn.

For each project include:

* Project name
* Objective
* Skills used
* What the learner should build
* Difficulty level

### 5. Milestones

Define measurable milestones throughout the roadmap so the learner can track their progress.

### 6. Assessment

Explain how the learner should test themselves at the end of each phase.

Include:

* Questions/concepts to test
* Practical tasks
* Project-based assessment
* Criteria for deciding whether they are ready to move forward

### 7. Resources

Suggest useful types of resources such as:

* Documentation
* Tutorials
* Books
* Courses
* Practice platforms
* Projects

Prioritize resources appropriate for the learner's current level.

### 8. Common Mistakes

List common mistakes learners make when studying this subject and explain how to avoid them.

### 9. Final Goal Preparation

Explain exactly what the learner should be capable of doing by the end of the roadmap to achieve their stated goal.

### 10. Final Checklist

Create a checklist of skills and concepts the learner should have mastered by the end.

**Important instructions:**

* Make the roadmap realistic and achievable within the specified time.
* Start from the learner's current level; do not unnecessarily repeat things they already know.
* Progress from fundamentals → intermediate concepts → advanced concepts → real-world application.
* Prioritize practical learning over passive theory.
* Include hands-on practice throughout the roadmap.
* Clearly distinguish between learning, practicing, and building.
* If the available time is limited, prioritize the most important topics.
* Adjust the depth and difficulty according to the learner's goal.
* Use clear headings, tables, bullet points, and numbered steps.
* Be specific rather than giving vague advice.
* The final roadmap should be detailed enough that the learner can follow it without needing to design their own study plan.

"""
)
```

```python
A='gsk_5OJhDGBh2GBJfX9Oe'
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.7,
    api_key=A
)
```

```python
chain1 = prompt1 | llm 
```

```python
result = chain1.invoke({
    "SUB": SUB,
    "LEV": LEV,
    "TIME": TIME,
    "DUR": DUR,
    "GOAL": GOAL
})
```

```python
print(result.content)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser

parser = JsonOutputParser()

prompt = PromptTemplate(input_variables=["text"],
    template="""
From the following text, extract the most relevant information
and create question-answer pairs.

TEXT:
{text}

Return JSON in this format:

{{
    "questions": [
        {{
            "question": "string",
            "answer": "string"
        }}
    ]
}}

{format_instructions}
""",
    
    partial_variables={
        "format_instructions": parser.get_format_instructions()
    }
)

qa_chain = prompt | llm | parser
```

```python
text_output = qa_chain.invoke({'text':result.content})
```

```python
text_output
```

```python
for qa in text_output["questions"]:
    print("Q:", qa["question"])
    print("A:", qa["answer"])
    print()
```
